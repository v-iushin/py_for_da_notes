```python
import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
```

```python
s = pd.Series(np.random.standard_normal(10).cumsum(),
              index=np.arange(0, 100, 10))
```

```python
s.plot()
```

```python
df = pd.DataFrame(np.random.standard_normal((10, 4)).cumsum(0),
                  columns=['A', 'B', 'C', 'D'],
                  index=np.arange(0, 100, 10))
```

```python
plt.style.use('grayscale')

df.plot()
```
