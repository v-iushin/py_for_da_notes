```python
import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
```

```python
fig = plt.figure()

ax = fig.add_subplot()
ax.plot(np.random.standard_normal(30).cumsum(),
        color='black',
        linestyle='--',
        marker='o')
```

```python
fig = plt.figure()

ax = fig.add_subplot()

data = np.random.standard_normal(30).cumsum()

ax.plot(data, color='black', linestyle='dashed', 
        label='Default')
ax.plot(data, color='black', linestyle='dashed',
        drawstyle='steps-post', label='steps-post')

ax.legend()
```
