```python
# numpy ufuncs also work
# with pandas objects
```

```python
import numpy as np
import pandas as pd
```

```python
frame = pd.DataFrame(np.random.standard_normal((4, 3)),
                     columns=list('bde'),
                     index=['Utah', 'Ohio', 'Texas', 'Oregon'])
frame
```

```python
np.abs(frame)
```

```python
# .apply() method
# apply function on 1-d array
```

```python
def f1(x):
    return x.max() - x.min()
```

```python
frame.apply(f1)
```

```python
frame.apply(f1, axis='columns')
```

```python
def f2(x):
    return pd.Series([x.min(), x.max()],
                     index=['min', 'max'])
```

```python
frame.apply(f2)
```

```python
def my_format(x):
    return f'{x:.2f}'
```

```python
# .map() method
# apply function on 1 scalar
```

```python
frame.map(my_format)
```

```python
frame['e'].map(my_format)
```
