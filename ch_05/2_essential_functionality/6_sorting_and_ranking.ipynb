```python
# sorting

# .sort_index() method
```

```python
import numpy as np
import pandas as pd
```

```python
obj = pd.Series(np.arange(4),
                index=['d', 'a', 'b', 'c'])
obj
```

```python
obj.sort_index()
```

```python
frame = pd.DataFrame(np.arange(8).reshape((2, 4)),
                     index=['three', 'one'],
                     columns=['d', 'a', 'b', 'c'])
frame
```

```python
frame.sort_index()
```

```python
frame.sort_index(axis='columns')
```

```python
frame.sort_index(axis='columns', ascending=False)
```

```python
# .sort_values() method
```

```python
obj = pd.Series([4, 7, -3, 2])
obj
```

```python
obj.sort_values()
```

```python
obj = pd.Series([4, np.nan, 7, np.nan, -3, 2])
obj
```

```python
obj.sort_values()
```

```python
obj.sort_values(na_position='first')
```

```python
frame = pd.DataFrame({'b': [4, 7, -3, 2],
                      'a': [0, 1, 0, 1]})
frame
```

```python
frame.sort_values('b')
```

```python
frame.sort_values(['a', 'b'])
```

```python
frame.sort_values(['b', 'a'])
```

```python
# ranking

# .rank() method
```

```python
obj = pd.Series([7, -5, 7, 4, 2, 0 ,4])
obj
```

```python
obj.rank()

# method='average' (default)
#
# element 4: index 4 and index 5
#       (4 + 5) / 2 = 4.5
# element 7: index 6 and index 7
#       (6 + 7) / 2 = 6.5
```

```python
obj.rank(method='first')
```

```python
obj.rank(ascending=False)
```

```python
frame = pd.DataFrame({'b': [4.3, 7, -3, 2],
                      'a': [0, 1, 0, 1],
                      'c': [-2, 5, 8, -2.5]})
frame
```

```python
frame.rank(axis='columns')
```
