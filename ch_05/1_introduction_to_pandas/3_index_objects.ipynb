```python
# Index objects
```

```python
import numpy as np
import pandas as pd
```

```python
# Index for Series
```

```python
obj = pd.Series(np.arange(3), index=['a', 'b', 'c'])
```

```python
index = obj.index
index
```

```python
index[1:]
```

```python
index[1] = 'd'
```

```python
labels = pd.Index(np.arange(3))
labels
```

```python
obj2 = pd.Series([1.5, -2.5, 0], index=labels)
obj2
```

```python
obj2.index is labels
```

```python
# Index for DataFrame
```

```python
populations = {
    'Ohio': 
        {
            2000: 1.5,
            2001: 1.7,
            2002: 3.6 
        },
    'Nevada':
        {
            2001: 2.4,
            2002: 2.9
        }
}
frame3 = pd.DataFrame(populations)

frame3.index.name = 'year'
frame3.columns.name = 'state'

frame3
```

```python
frame3.columns
```

```python
frame3.index
```

```python
'Ohio' in frame3.columns
```

```python
2003 in frame3.index
```
