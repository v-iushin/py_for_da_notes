```python
# nested dictionary of dictionaries

# outer keys - column indices
# inner keys - row indices
```

```python
populations = {
    'Ohio': 
        {
            2000: 1.5,
            2001: 1.7,
            2002: 3.6 
        },
    'Nevada':
        {
            2001: 2.4,
            2002: 2.9
        }
}
```

```python
frame3 = pd.DataFrame(populations)
frame3
```

```python
frame3.T
```

```python
pd.DataFrame(populations, index=[2001, 2002, 2003])
```

```python
# dictionary of series
```

```python
pdata = {
    'Ohio': frame3['Ohio'][:-1],
    'Nevada': frame3['Nevada'][:2]
}
```

```python
pd.DataFrame(pdata)
```

```python
frame3.index.name = 'year'
frame3.columns.name = 'state'

frame3
```
