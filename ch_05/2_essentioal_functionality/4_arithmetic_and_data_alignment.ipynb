```python
# arithmetic methods with fill values
```

```python
df1 = pd.DataFrame(np.arange(12.).reshape((3, 4)),
                   columns=list('abcd'))
df2 = pd.DataFrame(np.arange(20.).reshape((4, 5)),
                   columns=list('abcde'))
```

```python
df2.loc[1, 'b'] = np.nan
```

```python
df1
```

```python
df2
```

```python
df1 + df2
```

```python
df1.add(df2, fill_value=0)
```

```python
df1.add(df2, fill_value=np.nan)
```

```python
1 / df1
```

```python
df1.rdiv(1)
```

```python
df1.reindex(columns=df2.columns, 
            fill_value=0)
```

```python
df1.reindex(columns=df2.columns, 
            index=df2.index, 
            fill_value=0)
```
