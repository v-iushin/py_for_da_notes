```python
import numpy as np
import pandas as pd
```

```python
obj = pd.Series(np.arange(4.), index=['a', 'b', 'c', 'd'])
obj
```

```python
# for Series

# AVOID BARE ELEMENT LOOKUP 
# ser[label]
#
# USE BARE ONLY FOR BOOLEAN MASK
# ser[mask]
#
# USE 
# ser.loc[label] (inclusive if slice) AND 
# ser.iloc[ind] (exclusive if slice)
```

```python
obj['b']
```

```python
obj.loc['b']
```

```python
obj.iloc[1]
```

```python
obj.iloc[2:4]
```

```python
obj.loc[['b', 'a', 'd']]
```

```python
obj.iloc[[1, 3]]
```

```python
obj[obj < 2]
```

```python
obj
```

```python
obj.loc['b':'c']
```

```python
obj.loc['b':'c'] = 5
obj
```

```python
# for DataFrame

# BARE 
# frame[col_name]
# frame[mask]
#
# ONE ARGUMENT for .loc / .iloc
# frame.loc[row_name] / .iloc[row_ind]
# frame.loc[row_slice] / .iloc[row_slice]
#
# TWO ARGUMENTS for .loc / .iloc
# frame.loc[row_label, column_label]
# frame.iloc[row_ind, column_ind]
```

```python
data = pd.DataFrame(np.arange(16).reshape(4, 4),
                    index=['Ohio', 'Colorado', 'Utah', 'New York'],
                    columns=['one', 'two', 'three', 'four'])
data
```

```python
data['two']
```

```python
data.loc[:, 'two']
```

```python
data[['three', 'one']]
```

```python
data.loc[:, ['three', 'one']]
```

```python
# special cases
```

```python
data[:2]
```

```python
data.iloc[:2]
```

```python
data.iloc[:2, :]
```

```python
data[data['three'] > 5]
```

```python
data.loc[data.loc[:, 'three'] > 5, :]
```

```python
data < 5
```

```python
data[data < 5] = 0
data
```
