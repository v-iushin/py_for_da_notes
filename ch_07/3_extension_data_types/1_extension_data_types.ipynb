```python
import numpy as np
import pandas as pd
```

```python
s = pd.Series([1, 2, 3, None])
s
```

```python
s.dtype
```

```python
s = pd.Series([1, 2, 3, None], 
              dtype=pd.Int64Dtype())
s
```

```python
s.isna()
```

```python
s.dtype
```

```python
s[3]
```

```python
s[3] is pd.NA
```

```python
s[3] is np.nan
```

```python
s = pd.Series([1, 2, 3, None], 
              dtype='Int64')
s
```

```python
s = pd.Series(['one', 'two', None, 'three'], 
              dtype=pd.StringDtype())
s
```

```python
# with DataFrames
```

```python
df = pd.DataFrame({'A': [1, 2, None, 4],
                   'B': ['one', 'two', 'three', None],
                   'C': [False, None, False, True]})
df
```

```python
df['A'] = df['A'].astype('Int64')
df['B'] = df['B'].astype('string')
df['C'] = df['C'].astype('boolean')

df
```
