```python
# detect outliers
```

```python
import numpy as np
import pandas as pd
```

```python
data = pd.DataFrame(np.random.standard_normal((1000, 4)))

data.describe()
```

```python
col = data[2]

col[col.abs() > 3]
```

```python
data[(data.abs() > 3).any(axis='columns')]
```

```python
# cap values
```

```python
data[data.abs() > 3] = np.sign(data) * 3

data.describe()
```
