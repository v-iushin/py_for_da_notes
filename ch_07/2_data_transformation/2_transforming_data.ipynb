```python
# .map() method
# apply function on 1 scalar

# .apply() method
# apply function on 1-d array
```

```python
import numpy as np
import pandas as pd
```

```python
data = pd.DataFrame({"food": ["bacon", "pulled pork", "bacon", 
                              "pastrami", "corned beef", "bacon", 
                              "pastrami", "honey ham", "nova lox"], 
                     "ounces": [4, 3, 12, 6, 7.5, 8, 3, 5, 6]})
data
```

```python
meat_to_animal = {
    "bacon": "pig",
    "pulled pork": "pig",
    "pastrami": "cow",
    "corned beef": "cow",
    "honey ham": "pig",
    "nova lox": "salmon"
}
```

```python
data['animal'] = data['food'].map(meat_to_animal)

data
```

```python
def get_animal(x):
    return meat_to_animal[x]

data['food'].map(get_animal)
```
