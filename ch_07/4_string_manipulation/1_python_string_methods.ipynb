```python
# .split() and .strip()
```

```python
val = 'a,b,  guido'

val.split(',')
```

```python
pieces = [x.strip() for x in val.split(',')]
pieces
```

```python
# .join()
```

```python
first, second, third = pieces

first + '::' + second + '::' + third 
```

```python
'::'.join(pieces)
```
