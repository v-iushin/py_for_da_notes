```python
# .dropna() for Series
```

```python
import numpy as np
import pandas as pd
```

```python
data = pd.Series([1, np.nan, 3.5, np.nan, 7])
```

```python
data.dropna()
```

```python
data[data.notna()]
```

```python
# .dropna() for DataFrame
```

```python
data = pd.DataFrame([[1., 6.5, 3.],
                    [1., np.nan, np.nan],
                    [np.nan, np.nan, np.nan],
                    [np.nan, 6.5, 3.]])
data
```

```python
data.dropna()
```

```python
data.dropna(how='all')
```

```python
data[4] = np.nan
data
```

```python
data.dropna(axis='columns', how='all')
```

```python
df = pd.DataFrame(np.random.standard_normal((7, 3)))
df
```

```python
df.iloc[:4, 1] = np.nan
df.iloc[:2, 2] = np.nan

df
```

```python
df.dropna()
```

```python
df.dropna(thresh=2)
```
