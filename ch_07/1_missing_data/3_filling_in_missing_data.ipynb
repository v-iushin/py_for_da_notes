```python
# .fillna()
```

```python
import numpy as np
import pandas as pd
```

```python
df = pd.DataFrame(np.random.standard_normal((7, 3)))

df.iloc[:4, 1] = np.nan
df.iloc[:2, 2] = np.nan

df
```

```python
df.fillna(0)
```

```python
df
```

```python
df.fillna({1: 0.5, 2: 0})
```

```python
# .ffill() and .bfill()
```

```python
df = pd.DataFrame(np.random.standard_normal((6, 3)))

df.iloc[2:, 1] = np.nan
df.iloc[4:, 2] = np.nan

df
```

```python
df.ffill()
```

```python
df
```

```python
df.ffill(limit=2)
```

```python
data = pd.Series([1., np.nan, 3.5, np.nan, 7])
data
```

```python
data.fillna(data.mean())
```
