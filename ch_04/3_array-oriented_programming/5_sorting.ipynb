```python
# sort() method
```

```python
import numpy as np
rng = np.random.default_rng(seed=12345)
```

```python
arr = rng.standard_normal(6)
arr
```

```python
arr.sort()
arr
```

```python
arr = rng.standard_normal((5, 3))
arr
```

```python
arr.sort(axis=0)
arr
# axis=0: each column
```

```python
arr.sort(axis=1)
arr
# axis=1: each row
```

```python
arr2 = np.array([5, -10, 7, 1, 0, -3])
```

```python
sorted_arr2 = np.sort(arr2)
sorted_arr2
```
