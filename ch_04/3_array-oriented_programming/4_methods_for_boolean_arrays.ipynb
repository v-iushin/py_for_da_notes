```python
# sum() method
```

```python
import numpy as np
rng = np.random.default_rng(seed=12345)
```

```python
arr = rng.standard_normal(100)
```

```python
arr > 0

# True = 1, False = 0
```

```python
(arr > 0).sum()
```

```python
(arr <= 0).sum()
```

```python
# any() and all() methods
```

```python
bools = np.array([False, False, True, False])
```

```python
bools.any()
```

```python
bools.all()
```

```python
print(bools.all())
```

```python
# can be used with non-Boolean arrays
# nonzero - True
# zero - False
```
