```python
import numpy as np
```

```python
points = np.arange(-5, 5, 0.01)
```

```python
xs, ys = np.meshgrid(points, points)

# x = np.array([1, 2, 3])
# y = np.array([10, 20])
#
# X, Y = np.meshgrid(x, y)
# 
# X
# -> 
# [[1, 2, 3],
#  [1, 2, 3]]
#
# Y
# ->
# [[10, 10, 10],
#  [20, 20, 20]]
#
# both X and Y have shape (len(y), len(x))
# this is indexing='xy' (default)
# to make shape (lex(x), len(y))
# indexing='ij'
```

```python
xs
```

```python
ys
```

```python
z = np.sqrt(xs ** 2 + ys ** 2)
z
```
