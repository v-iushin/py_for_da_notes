```python
# fancy indexing is
# indexing using integer array
```

```python
import numpy as np
```

```python
arr = np.zeros((8, 4))
```

```python
for i in range(8):
    arr[i] = i

arr
```

```python
# selecting subset
```

```python
arr[[4, 3, 0, 6]]

# 4-th row, 3-rd row, 0-th row, 6-th row
```

```python
arr[[-3, -5, -7]]

# 3-rd row from end, 5-th row from end, ...
```

```python
arr = np.arange(32).reshape((8, 4))
arr
```

```python
arr[[1, 5, 7, 2], [0, 3, 1, 2]]

# elements: (1,0), (5,3), (7,1), (2,2)
```

```python
arr[[1, 5, 7, 2]][:, [0, 3, 1, 2]]

# firstly selecting susbet 
# using integer array
#
# arr[[1, 5, 7, 2]] = a1
# ->
# [[4, 5, 6, 7],
#  [20, 21, 22, 23],
#  [28, 29, 30, 31],
#  [8, 9, 10, 11]]
#
# secondly selecting and rearranging columns
#
# a1[:, [0, 3, 1, 2]]
# ->
# [[4, 7, 5, 6],
#  [20, 23, 21, 22],
#  [28, 31, 29, 30],
#  [8, 11, 9, 10]]
#
# or differently
#
# [[(1,0), (1,3), (1,1), (1,2)],
#  [(5,0), (5,3), (5,1), (5,2)],
# ...
```

```python
# setting value
```

```python
arr[[1, 5, 7, 2], [0, 3, 1, 2]]
```

```python
arr[[1, 5, 7, 2], [0, 3, 1, 2]] = 0
arr
```
