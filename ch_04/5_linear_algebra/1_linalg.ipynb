```python
import numpy as np
rng = np.random.default_rng(seed=12345)
```

```python
x = np.array([[1., 2., 3.], [4., 5., 6.]])
y = np.array([[6., 23.], [-1, 7], [8, 9]])
```

```python
x
```

```python
y
```

```python
x.dot(y)
```

```python
np.dot(x, y)
```
