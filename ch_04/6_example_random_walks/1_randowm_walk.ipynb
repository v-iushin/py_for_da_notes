```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
import random
position = 0
walk = [position]
nsteps = 1000
for _ in range(nsteps):
    step = 1 if random.randint(0, 1) else -1
    position += step
    walk.append(position)
walk[:100]
```

```python
plt.plot(walk[:100])
```

```python
# numpy version
```

```python
nsteps = 1000
rng = np.random.default_rng(seed=12345)
draws = rng.integers(0, 2, size=nsteps)
steps = np.where(draws == 0, 1, -1)
walk = steps.cumsum()
walk[:100]
```

```python
plt.plot(walk[:100])
```

```python
walk.min()
```

```python
walk.max()
```

```python
# check when first time random walk
# got 10 steps away

(np.abs(walk) >= 10).argmax()
```
