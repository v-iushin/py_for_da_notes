```python
# creating dictionaries from sequencess
```

```python
mapping = {}
key_list = ['k1', 'k2', 'k3']
value_list = ['v1', 'v2', 'v3']

for key, value in zip(key_list, value_list):
    mapping[key] = value

mapping
```

```python
tuples = zip(range(5), reversed(range(5)))
tuples
```

```python
mapping = dict(tuples)
mapping
```

```python
# default values
```

```python
words = ['apple', 'bat', 'bar', 'atom', 'book']
by_letter = {}

for word in words:
    letter = word[0]
    if letter not in by_letter:
        by_letter[letter] = [word]
    else:
        by_letter[letter].append(word)

by_letter
```

```python
words = ['apple', 'bat', 'bar', 'atom', 'book']
by_letter = {}

for word in words:
    letter = word[0]
    by_letter.setdefault(letter, []).append(word)

by_letter
```

```python
from collections import defaultdict

words = ['apple', 'bat', 'bar', 'atom', 'book']

by_letter = defaultdict(list)

for word in words:
    by_letter[word[0]].append(word)

by_letter
```
