```python
# set operations
```

```python
a = {1, 2, 3, 4, 5}
b = {3, 4, 5, 6, 7, 8}
```

```python
a.union(b)
```

```python
a | b
```

```python
a.intersection(b)
```

```python
a & b
```

```python
# in-place counterparts
```

```python
c = a.copy()
c |= b
c
```

```python
d = a.copy()
d &= b
d
```

```python
# subset and superset
```

```python
a_set = {1, 2, 3, 4, 5}
```

```python
{1, 2, 3}.issubset(a_set)
```

```python
a_set.issuperset({1, 2, 3})
```
