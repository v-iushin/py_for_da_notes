```python
# zip
```

```python
seq1 = ['foo', 'bar', 'baz']
seq2 = ['one', 'two', 'three']

zipped = zip(seq1, seq2)
list(zipped)
```

```python
seq3 = [False, True]
list(zip(seq1, seq2, seq3))
```

```python
for index, (a, b) in enumerate(zip(seq1, seq2)):
    print(f'{index}: {a}, {b}')
```
