```python
path = '../../../_data_py_for_da/examples/segismundo.txt'
```

```python
f = open(path, encoding='utf-8')
```

```python
for line in f:
    print(line)
```

```python
f.seek(0)
```

```python
lines = [x.rstrip() for x in f]
lines
```

```python
f.close()
```

```python
# with statement
```

```python
with open(path, encoding='utf-8') as f:
    lines = [x.rstrip() for x in f]

lines
```
