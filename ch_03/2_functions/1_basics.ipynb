```python
def my_function(x, y):
    return x + y

my_function(1, 2)
```

```python
result = my_function(1, 2)
result
```

```python
def function_without_return(x):
    print(x)

result = function_without_return('hello!')
```

```python
print(result)
```

```python
def my_function2(x, y, z=1.5):
    if z > 1:
        return z * (x + y)
    else:
        return z / (x + y)
```

```python
my_function2(5, 6, z=0.7)
```

```python
my_function2(3.14, 7, 3.5)
```

```python
my_function2(10, 20)
```

```python
# returning multiple values
```

```python
def f():
    a = 5
    b = 6
    c = 7
    return a, b, c

return_value = f()
return_value
```

```python
a, b, c = f()
a, b, c
```

```python
def f():
    a = 5
    b = 6
    c = 7
    return {'a': a, 'b': b, 'c': c}

f()
```
