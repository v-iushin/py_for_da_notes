```python
def attempt_float(x):
    try:
        return float(x)
    except:
        return x
```

```python
attempt_float('1.2345')
```

```python
attempt_float('something')
```

```python
def attempt_float(x):
    try:
        return float(x)
    except ValueError:
        return x
```

```python
attempt_float((1, 2))
```
