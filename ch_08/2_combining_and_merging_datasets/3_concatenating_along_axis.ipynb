```python
# numpy.concatenate()
```

```python
import numpy as np
import pandas as pd
```

```python
arr = np.arange(12).reshape((3, 4))
arr
```

```python
np.concatenate([arr, arr], axis=1)
```

```python
# pandas.concat()
# pandas.concat(axis='index') - default
```

```python
s1 = pd.Series([0, 1], index=['a', 'b'], dtype='Int64')

s2 = pd.Series([2 ,3, 4], index=['c', 'd', 'e'], dtype='Int64')

s3 = pd.Series([5, 6], index=['f', 'g'], dtype='Int64')
```

```python
s1
```

```python
s2
```

```python
s3
```

```python
pd.concat([s1, s2, s3])
```

```python
pd.concat([s1, s2, s3], axis='columns')
```

```python
# pandas.concat(join='inner')
# pandas.concat(join='outer') - default
```

```python
s4 = pd.concat([s1, s3])
s4
```

```python
pd.concat([s1, s4], axis='columns')
```

```python
pd.concat([s1, s4], axis='columns', 
          join='inner')
```

```python
# pandas.concat(keys=)
```

```python
result = pd.concat([s1, s1, s3], 
                   keys=['one', 'two', 'three'])
result
```

```python
result.unstack()
```

```python
pd.concat([s1, s2, s3], axis='columns',
          keys=['one', 'two', 'three'])
```

```python
# for DataFrames
```

```python
df1 = pd.DataFrame(np.arange(6).reshape(3, 2),
                   index=['a', 'b', 'c'],
                   columns=['one', 'two'])

df2 = pd.DataFrame(2 + np.arange(4).reshape(2, 2),
                   index=['a', 'c'],
                   columns=['three', 'four'])
```

```python
df1
```

```python
df2
```

```python
pd.concat([df1, df2], axis='columns',
          keys=['level1', 'level2'])
```

```python
pd.concat({'level1': df1, 'level2': df2},
          axis='columns')
```

```python
# pandas.concat(names=)
```

```python
pd.concat([df1, df2], axis='columns',
          keys=['level1', 'level2'],
          names=['upper', 'lower'])
```

```python
# pandas.concat(ignore_index=True)
```

```python
df1 = pd.DataFrame(np.random.standard_normal((3, 4)),
                   columns=['a', 'b', 'c', 'd'])

df2 = pd.DataFrame(np.random.standard_normal((2, 3)),
                   columns=['b', 'd', 'a'])
```

```python
df1
```

```python
df2
```

```python
pd.concat([df1, df2], ignore_index=True)
```
