```python
import numpy as np
import pandas as pd
```

```python
path = '../../../_data_py_for_da/'
```

```python
data = pd.read_csv(path+'examples/macrodata.csv')
```

```python
data.head()
```

```python
data = data.loc[:, ['year', 'quarter', 'realgdp', 'infl', 'unemp']]
data.head()
```

```python
periods = pd.PeriodIndex.from_fields(year=data.pop('year'), 
                                     quarter=data.pop('quarter'))
periods.name = 'date'

periods
```

```python
data.index = periods.to_timestamp('D')

data.head()
```

```python
data = data.reindex(columns=['realgdp', 'infl', 'unemp'])
data.columns.name = 'item'

data.head()
```

```python
long_data = (data.stack()
             .reset_index()
             .rename(columns={0: 'value'})
             )

long_data[:10]
```

```python
# long to wide

# .pivot()
```

```python
pivoted = long_data.pivot(index='date',
                          columns='item',
                          values='value')

pivoted.head()
```
