```python
# .stack() and .unstack()
```

```python
import numpy as np
import pandas as pd
```

```python
data = pd.DataFrame(np.arange(6).reshape(2, 3),
                    index=pd.Index(['Ohio', 'Colorado'], name='state'),
                    columns=pd.Index(['one', 'two', 'three'], name='number'))
data
```

```python
result = data.stack()
result
# .stack() - columns -> rows
# -> hierarchical
```

```python
result.unstack()
# .unstack() - rows -> columns
# .unstack(level=-1) - default
```

```python
result.unstack(level=0)
```

```python
result.unstack(level=1)
```

```python
result.unstack(level='state')
```

```python
s1 = pd.Series([0, 1, 2, 3],
               index=['a', 'b', 'c', 'd'],
               dtype='Int64')

s2 = pd.Series([4, 5, 6],
               index=['c', 'd', 'e'],
               dtype='Int64')
```

```python
data2 = pd.concat([s1, s2], keys=['one', 'two'])
data2
```

```python
data2.unstack()
```

```python
data2.unstack().stack()
```

```python
data2.unstack().stack().dropna()
```
