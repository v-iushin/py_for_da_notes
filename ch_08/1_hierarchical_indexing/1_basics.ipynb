```python
import numpy as np
import pandas as pd
```

```python
data = pd.Series(np.random.uniform(size=9),
                 index=[['a', 'a', 'a',
                         'b', 'b',
                         'c', 'c',
                         'd', 'd'],
                        [1, 2, 3, 1, 3, 1, 2, 2, 3]])
data
```

```python
data.index
```

```python
data['b']
```

```python
data.loc['b':'c']
```

```python
data.loc[['b', 'd']]
```

```python
data.loc[:, 2]
```

```python
# .unstack() and .stack()
```

```python
data.unstack()
```

```python
data.unstack().stack()
```

```python
data.unstack().stack().dropna()
```

```python
# for DataFrame
```

```python
frame = pd.DataFrame(np.arange(12).reshape((4, 3)),
                     index=[['a', 'a', 'b', 'b'],
                            [1, 2, 1, 2]],
                     columns=[['Ohio', 'Ohio', 'Colorado'],
                              ['Green', 'Red', 'Green']])
frame
```

```python
frame.index.names = ['key1', 'key2']

frame.columns.names = ['state', 'color']

frame
```

```python
frame.index.nlevels
```

```python
frame['Ohio']
```

```python
mi = pd.MultiIndex.from_arrays([['Ohio', 'Ohio', 'Colorado'],
                          ['Green', 'Red', 'Green']],
                          names=['state', 'color'])

frame1 = pd.DataFrame(np.arange(12).reshape((4, 3)),
                      index=[['a', 'a', 'b', 'b'],
                             [1, 2, 1, 2]],
                      columns=mi)
frame1
```
